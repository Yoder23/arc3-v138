````python
import os, time, urllib.request, signal
if os.getenv('KAGGLE_IS_COMPETITION_RERUN'):
    ready_until=time.monotonic()+600
    while True:
        try:
            with urllib.request.urlopen('http://gateway:8001/api/games',timeout=10) as response:
                assert response.status==200
            break
        except Exception:
            if time.monotonic()>=ready_until:raise RuntimeError('Competition gateway unavailable')
            time.sleep(5)
    process=subprocess.Popen([sys.executable,'-B',str(root/'client.py')],start_new_session=True)
    try:
        process.wait(timeout=30000)
    except subprocess.TimeoutExpired:
        os.killpg(process.pid,signal.SIGKILL)
        process.wait()
        raise RuntimeError('Whole-notebook agent budget exhausted')
    if process.returncode:raise RuntimeError('Agent driver failed')
else:
    # Official starter convention: schema fixture for commit mode only.
    # Real scores/submission.parquet are emitted by the gateway on competition rerun.
    import pandas as pd
    pd.DataFrame([['1_0','1',True,0]],columns=['row_id','game_id','end_of_game','score']).to_parquet('/kaggle/working/submission.parquet',index=False)
    print('Commit-only schema fixture; no games evaluated and no score claimed.')
````
